```python
plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"])
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()
```

```python
# replacing species column with numbers
iris["Species"] = iris["Species"].replace({"Iris-setosa":3,"Iris-versicolor":4,"Iris-virginica":5})
iris.sample(4)
```

```python
# c (color) --> parameter
# cmap (colormap) ---> check internet for more --> jet ,summer, winter , viridis--> color styles
# plt.colorbar() -->  display colorbar for indentifying the datapoint/category
# alpha parameter --> opacity of colors
plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"],c=iris["Species"],cmap="jet",alpha= 0.7)
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.colorbar()
```

**Plot Size**

```python
# plt.figure() -> figsize= (width, height) ---> for plot sizing (any plot)

plt.figure(figsize=(15,7))

plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"],c=iris["Species"],cmap="jet",alpha= 0.7)
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.colorbar()
plt.show()
```

```python
# using subplots to plot two different plots simultaneouly
# each cell in subplot represent the one plot
# nrows * ncols = no. of plots
# sharex parameter
fig, ax = plt.subplots(nrows=2,ncols=1,figsize=(12,10),sharex=True)
# fig --> figure
# ax ---> array of axes
ax[0].scatter(df["avg"],df["strike_rate"],color="red")
ax[1].scatter(df["avg"],df["runs"])

ax[0].set_title("Analysis of Average & Strike Rate")
ax[0].set_ylabel("Strike Rate")

ax[1].set_title("Analysis of  Runs & Average")
ax[1].set_xlabel("Average")
ax[1].set_ylabel("Runs")

plt.savefig("Subplot")
```

## **3D Surface Plots --> Widely Used**

```python
x= np.linspace(-20,20,100)
y= np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)
```

```python
z= xx**2 + yy**2

```

```python
fig = plt.figure(figsize= (15,10))

ax= plt.subplot(projection="3d")
p = ax.plot_surface(xx,yy,z,cmap="viridis")
# color bar ---> store plot in p variable then use in fig.colorbar() function --> it uses z value
fig.colorbar(p)
plt.show()
```

```python
x= np.linspace(-30,30,100)
y = np.linspace(-20,20,100)
xx,yy = np.meshgrid(x,y)
z= 4*(xx**2) / np.sqrt(3*yy) +2

fig = plt.figure(figsize= (15,10))

ax= plt.subplot(projection="3d")
p = ax.plot_surface(xx,yy,z,cmap="viridis")

fig.colorbar(p)
plt.show()
```

```python
x= np.linspace(-30,30,100)
y = np.linspace(-20,20,100)
xx,yy = np.meshgrid(x,y)
z= 4*(xx**2) * 9*(yy**2) +2

fig = plt.figure(figsize= (15,10))

ax= plt.subplot(projection="3d")
p = ax.plot_surface(xx,yy,z,cmap="viridis")

fig.colorbar(p)
plt.show()
```

```python
z= np.tan(xx) - np.sin(yy)

fig = plt.figure(figsize= (15,10))

ax= plt.subplot(projection="3d")
p = ax.plot_surface(xx,yy,z,cmap="viridis")

fig.colorbar(p)
plt.show()
```

```python
z= np.cos(xx) + np.log(yy)

fig = plt.figure(figsize= (15,10))

ax= plt.subplot(projection="3d")
p = ax.plot_surface(xx,yy,z,cmap="viridis")

fig.colorbar(p)
plt.show()
```

## Contour plot

```python
# explain the 3d plot in 2d using colors
x= np.linspace(-10,10,100)
y= np.linspace(-10,10,100)
xx,yy = np.meshgrid(x,y)
z= xx**2 + yy**2 +8

fig = plt.figure(figsize= (12,10))
ax= plt.subplot()
ax.contour(x,y,z,cmap="viridis")
```

```python
# contourf ---> filled color
x= np.linspace(-10,10,100)
y= np.linspace(-10,10,100)
xx,yy = np.meshgrid(x,y)
z= xx**2 + yy**2 +8

fig = plt.figure(figsize= (12,10))
ax= plt.subplot()
ax.contourf(x,y,z,cmap="viridis")
```
