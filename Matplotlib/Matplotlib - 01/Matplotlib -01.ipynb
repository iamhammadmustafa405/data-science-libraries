```python
# the problem ---> when doing this the next plot is coming infront of the last plot thats why it is not useful
plt.bar(df["batsman"],df["2015"])
plt.bar(df["batsman"],df["2016"])
plt.bar(df["batsman"],df["2017"])
```

```python
df
```

```python
# the solution -> there is no direct method for dealing this problem in the matplotlin library, but we can handle this in this way;
# judgar using width parameter
# by using number and shifting by subtracting and adding instead of name of player
# this is simple calculation  we have to set width and shift all data of   each category combined and seperate from other category



plt.bar(np.arange(df.shape[0])-0.2,df["2015"],width=0.2,color="black",label="2015")
plt.bar(np.arange(df.shape[0]),df["2016"],width=0.2,color="blue",label="2016")
plt.bar(np.arange(df.shape[0]) + 0.2,df["2017"],width=0.2, color= "red",label="2017")

# but there are numbers instead of names of player in x-axis
# plt.xticks(old_set,new_set)
plt.title("Analysis of IPL Top 5 Batsman 2015 - 2017 Performance")
plt.xlabel("Players")
plt.ylabel("Runs")
plt.xticks(np.arange(df.shape[0]),df["batsman"])
plt.legend()
plt.show()
```

```python
# a problem ---> when name of category is to long ---> readability is zero
children= [15,8,4,10]
colors = ['redredredredredredredred','greengreengreengreengreengreen','pinkpinkpinkpinkpink','whitewhitewhitewhitewhitewhitewhitewhite']

plt.bar(colors,children,color='black')
plt.title("Colors Analysis")
plt.xlabel("Colors")
plt.ylabel("Children")
```

```python
# sol ---> plt.xticks(rotation="vertical") or any angle given so the x ticks are rotate by that angle and readability can be enhanced.

children= [15,8,4,10]
colors = ['redredredredredredredred','greengreengreengreengreengreen','pinkpinkpinkpinkpink','whitewhitewhitewhitewhitewhitewhitewhite']

plt.bar(colors,children,color='black')
plt.title("Colors Analysis")
plt.xlabel("Colors")
plt.ylabel("Children")
plt.xticks(rotation="vertical")
```

```python
# stacked bar chart ---> bottom parameter --> means column bottom is attached to which column?
plt.bar(df["batsman"],df["2015"],label="2015")
plt.bar(df["batsman"],df["2016"],bottom=df["2015"],label="2016")
plt.bar(df["batsman"],df["2017"],bottom=(df["2015"] + df["2016"]),label="2017") # doing sum of previous two columns 

plt.legend()
plt.show()
```

```python
# for percentage contribution of each category --> autopct= "%0.1f%%" <-- parameter
# colors --> parameter
plt.pie(df["batsman_runs"], labels = df["batsman"],autopct="%0.1f%%", colors = ["blue","green","yellow","black","cyan","brown"])
plt.show()
```

```python
# explode --> explode parameter --> explode = [0,0,0,0,0.1] etc.
# for extracting a specific piece from the pie
# shadow parameter
plt.pie(df["batsman_runs"],labels=df["batsman"],autopct="%0.1f%%",explode = [0.1,0,0,0.5,0.1,0], shadow= True)
plt.show()
```
