```python
# import libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
# line chart 
#px.line(x,y,title)
names= ["Hammad","Huzaifa","Ammar"]
heights = [5.7,5.8,5.9]

fig = px.line(x=names,y=heights,title="Height of Students",)
fig.show()
```

```python
# bar --> px.bar()
# fig.show()
students= ["huzaifa","haider","hammad"]
candies = [19,7,2]

fig = px.bar(x= students, y= candies,title="No.of Candies Got Each Student")
fig.show()
```

```python
# save the plot in png file ---> fig.write_image(path + name, format, scale)
x= np.linspace(-10,10,20)
y= np.linspace(-10,5,20)

fig = px.line(x,y)
# save in 300 DPI ( Dots Per Inch) scale = 3

fig.write_image("Line Chart in Plotly", format="png", scale=3) # only be done when kaleido is installed
```

```python
# save as standalone html file ---> fig.write_html()
x= np.linspace(-10,10,20)
y= np.linspace(-10,5,20)

fig = px.line(x,y)
fig.write_html("Line Chart in Plotly.html")
```

## Heatmaps

```python
import plotly.graph_objects as go
# go.Heatmap(z= data_for heatmap) this Class object is used to make Heatmap of the data.

data = np.random.rand(10,10)

pd.DataFrame(data)
```

```python
# Heatmap of the Data




heatmap = go.Heatmap(z=data) # z parameter

# heatmap is the object of the class go.Heatmap(data)
# using this ( heatmap ) object  in the go.Figure class  we make another object (fig)  that object is heatmap.

fig = go.Figure(data=heatmap)
fig.show()
```

```python
# Heatmap on correlation corr of titanic dataset of age and fare using plotly
# using px.imshow() method
# 1st method
data = df[["Age","Fare"]].corr()
fig = px.imshow(data)
fig.show()
```

```python
# nticks --> x,yaxis
# size of plot using width and height in dictionary

data = df[["Age","Fare"]].corr()
heatmap = go.Heatmap(z=data)
fig = go.Figure(heatmap)
fig.update_layout(
  {
    "title": "Correlation between Age & Fare",
    "xaxis" : {"title": "Age","nticks": 50},
    "yaxis": {"title": "Fare","nticks":50},
    "width": 1000,
    "height": 800
  }
)
```
