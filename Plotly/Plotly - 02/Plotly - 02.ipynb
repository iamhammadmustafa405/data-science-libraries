# **15+ Plots in Plotly**


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
```

```python
# line chart
df = pd.read_csv(r"iris.csv")
df.head()
```

```python
# sorting
df_sort = df.sort_values(by = "SepalWidthCm")
# plotting

fig = px.line(df_sort,x="SepalWidthCm",y="PetalWidthCm",color = "Species",width=1200,height=800)
fig.write_html("./Plots/Line Chart.html")

```

```python
fig.show()
```

```python
df.head()
```

```python
# bar plot
# its necessary to group values based on means then plot
#
df_bar = df.groupby(["Species"]).mean().reset_index()
df_bar
```

```python
fig = px.bar(df_bar,x="Species",y="SepalLengthCm",)
fig.show()
```

## BoxPlot

```python
# boxplot
fig = px.box(df,x="Species",y="SepalLengthCm",color = "Species")
fig.show()

fig.write_html("./Plots/Boxplot in Plotly.html")
```

## **Violin Plot ---> Very Interactive**

```python
# violon plot --> kde --> expension of the graph
fig = px.violin(df,x="Species",y="SepalWidthCm",color="Species")
fig.show()

```

```python
# box plot in violin --> boz = True Parameter
fig = px.violin(df,x="Species",y="SepalWidthCm",color="Species",box = True)
fig.show()

fig.write_html("./Plots/Violin Plot in Plotly.html")
```

## **Histogram**

```python
#histogram
vk = pd.read_csv(r"vk.csv")
vk.head()
```

```python
# px.histogram()
# nbins parameter ---> no. of bins
fig = px.histogram(vk,x="batsman_runs", nbins=10,title="Virat Kohli IPL Career Analysis")
fig
```

## **Piechart**

```python
# piechart
df_pie = df["Species"].value_counts().reset_index()
df_pie
```

```python
# px.pie(df,values,names)
fig = px.pie(df_pie,values="count",names="Species")
fig.show()
```

## **3D Scatter Plot**

```python
# px.scatter_3d
fig = px.scatter_3d(df,x='SepalLengthCm',y='SepalWidthCm', z= 'PetalWidthCm', color = "Species")
fig.show()
fig.write_html("./Plots/ 3D Scatter Plot in plotly.html")
```

## **Area Chart**

```python
df
```

```python
# area chart

# px.area()
fig = px.area(df, x= "SepalWidthCm", y = "SepalLengthCm", color = "Species",title="Iris Analysis")
fig.show()
fig.write_html("./Plots/Area chart.html")
```

## **Bubble Chart**

#

```python
# simple scatter but when size parameter is used
fig = px.scatter(df,x ="SepalLengthCm", y="SepalWidthCm", size="PetalWidthCm",title ="Bubble Chart")
fig.show()
fig.write_html("./Plots/Bubble Chart.html")
```

## **Sunburst Chart --> Best Interaction**

```python
# when primary categories has more categories then it is used
# path and values  parameter
df_sunburst = df.groupby(["Species","PetalWidthCm"]).size().reset_index(name="Counts") # creating subgroups in a group

fig = px.sunburst(df_sunburst, path=["Species","PetalWidthCm"], values = "Counts")
fig.show()
```

```python
# Practice -> Analysis of IPL Playoffs & Finals in Different Cities
ipl = pd.read_csv(r"ipl-matches.csv")
ipl.head(2)

```

```python
ipl = ipl[~(ipl["MatchNumber"].str.isdigit())]
host_match= ipl.groupby(["City","MatchNumber"])["MatchNumber"].count().reset_index(name="No. of Matches")
host_match.head()
```

```python
fig= px.sunburst(host_match,path=["City","MatchNumber"],values="No. of Matches",title="Analysis of IPL Playoffs & Finals in Different Cities")
fig.show()
fig.write_html("./Plots/Analysis of IPL Playoffs & Finals in Different Cities.html")
```

## **Paralllel Coordinates Plot**

```python
df.columns
```

```python
# px.parallel_coordinates()
# removing id column and replacing Species categories with numbers for color parameter in function.
df_parallel = df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm','Species']]

# another method of converting the category species in catcodes like 0,1,2 etc.

df_parallel["Species"] = df_parallel["Species"].astype("category").cat.codes
df_parallel.sample(5)
```

```python
# color, labels --> parameter
fig  = px.parallel_coordinates(df_parallel,color="Species",labels = {"Species":"Kinds"},width=1000,height=500)
fig.show()
fig.write_html("./Plots/Parallel Coordinates plot.html")
```

## **Density Contour Plots**

```python
# px.density_contour()
fig = px.density_contour(df,x="SepalLengthCm", y="SepalWidthCm", color="Species")
fig.show()
fig.write_html("./Plots/Density Contour Plot.html")
```

## **Ternary plot**

```python
# px.scatter_ternary() --> plot in Triangle --> a,b,c, color parameter
# use for finding outliers in the data
fig = px.scatter_ternary(df,a='SepalLengthCm',b='SepalWidthCm',c='PetalWidthCm', color="Species")
fig.show()
fig.write_html("./Plots/Scatter Ternary Plot.html")
```

## **Polar Chart (Radar Chart)**

```python
# px.line_polar ---> r, theta, line_close parameters
# usage:  give difference of different Categories on the basis of specific column
df_radar = df.groupby(["Species"]).mean().reset_index()
df_radar
```

```python
fig = px.line_polar(df,r="PetalWidthCm",theta = "Species", line_close= True)
fig.show()
fig.write_html("./Plots/Polar (Radar) Chart.html")
```
