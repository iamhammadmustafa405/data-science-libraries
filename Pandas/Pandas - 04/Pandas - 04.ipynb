## Groupby Attributes & Methods

```python
# find total number of groups --> len
len(movies.groupby("Genre")) # gives how many groups are made of genres means how many different types of genres are there
```

```python
# using nunique to get:
movies["Genre"].nunique()
```

```python
# how many items in each group ---> size() --> ( gives result in series)
movies.groupby("Genre").size() # ---> give the no. of movies in each Genre(Category/Group) ---> prefer this
```

```python
# same result also by:
movies["Genre"].value_counts() # it is sorted
```

```python
# first() ---> fetch the first movie of each group in a dataframe
genres= movies.groupby("Genre") # ---> genres is a groupby object
genres.first() # ----> it gives the first movie of each Genre(Group).
```

```python
# last() ---> fetch the last movie of each group in a dataframe
genres.last() # ----> it gives the last movie of each Genre(Group).
```

```python
# nth() ---> give the nth(index) movie of each group in a dataframe.
genres.nth(6) # ---> give the 6th index (7th movie) of each Genre(Group)
# Note: Many groups are not there because many groups haven't 6th index movies they have movies lower than that index. 
```

```python
# get_group() --->fetching a particular group.
genres.get_group("Adventure") # ---> faster than boolean masking and filtering
```

```python
# by using filtering:
movies[movies["Genre"] =="Adventure"]
```

```python
# groups --->attribute
# gives a dictionary in which keys are groups and values are lists in which the group items indices are stored.
genres.groups
```

```python
# describe() --> gives the high mathematical information of all groups in a dataframe (same as in series and dataframe)
genres.describe()
```

```python
# sample() # ---> randomly give item (movies) from each group (default 1)
# replace=True parameter ---> means if we select 2 and a group has only 1 it repeats the 1 and complete the 2
genres.sample(2,replace = True) # fetch 2 movies from each group
```

```python
# nunique ---> gives the no. of unique values for each column in the dataframe for groupby object
genres.nunique()
```

```python
genres.sum(numeric_only=True)
```

```python
# agg() method ---> passing dict --> apply different aggregate functions on differnt columns by passing a dictionary.
# {"column_name":"aggregate_method_name"}

# condition 01: applying different but one aggregate function to each column.
genres.agg(
    {
        "Runtime":"mean",
        "IMDB_Rating":"mean",
        "No_of_Votes": "sum",
        "Gross":"sum",
        "Metascore":"mean"
    }
)
```

```python
# Condition 02: applying multiple functions to all  columns
genres.agg(
    {
    "Runtime":["sum","min","max","mean"],
        "IMDB_Rating":["sum","min","max","mean"],
        "No_of_Votes": ["sum","min","max","mean"],
        "Gross":["sum","min","max","mean"],
        "Metascore":["sum","min","max","mean"]
}
)
# by using this we can get all lot of information about the data.
```

```python
# Looping on groups
# "group" is the name of the group in a str datatype and "data" is the dataframe of each group. i.e. action movies has its own dataframe and adventure has its own by looping it we can get all groups --> dataframe in data variable

for group,data in genres: 
    print(group) # group name
    print(data) # group dataframe
```

## Split Apply Combine Strategy

```python
# split ---> split the groups using groupy
#apply ---> apply any logic to all the groups and then #combine
```

```python
# apply() --> can be used to apply any function: builtin or custom on groups.
movies.groupby("Genre")["No_of_Votes"].apply(max)
```

```python
# find number of movies starting with "a" for each group

genres # --> spliting the dataframe into groups by groupBy/
def foo(group): # ----> applying custom logic to every group
    return group["Series_Title"].str.startswith("A").sum()  # type of group is dataframe
    


genres.apply(foo) # --> then again combine
# This is Split Apply Combine Strategy.
```

```python
# 2: find ranking of each movie in the group according to the IMDB Score

def rank_movie(group):
    group["Genre_Rank"] = group["IMDB_Rating"].rank(ascending = False)
    return group

genres.apply(rank_movie)
```

```python
# 3: find normalized rating (IMDB) for each movie group wise.

def norm(group):
    group["Norm_rating"] =(group["IMDB_Rating"] - group["IMDB_Rating"].min())/(group["IMDB_Rating"].max() -group["IMDB_Rating"].min())
    group["Norm_rating"] = np.round(group["Norm_rating"],2)
    return group # this will return the all groups dataframe not the original one
```

```python
genres.apply(norm)
```

# Exercise

```python
ipl= pd.read_csv(r"deliveries.csv")
ipl.head(20)
```

```python
# 1: find the top 10 batsman in terms of runs scored.
ipl.groupby("batsman")["batsman_runs"].sum().sort_values(ascending =False).head(10)
```

```python
# 2: find the batsman with max no of sixes
sixes = ipl[ipl["batsman_runs"] == 6]
sixes.groupby("batsman")["batsman_runs"].count().sort_values(ascending = False).head(1).index[0]
```

```python
# 3: find the batsman with most no of 4's and 6's in last 5 overs
filtered = ipl[(ipl["over"] > 15) & ((ipl["batsman_runs"]== 4) | (ipl["batsman_runs"] == 6))]
filtered.groupby("batsman")["batsman_runs"].value_counts().sort_values(ascending = False).head(2)
```

```python
# 4: find V Kohli's record against all teams.
temp_df = ipl[ipl["batsman"]== "V Kohli"]
temp_df.groupby("bowling_team")["batsman_runs"].sum().sort_values(ascending=False).reset_index()
```

```python
#5: find highest virat kohli score in ipl history
tem_df = ipl[ipl["batsman"] =="V Kohli"]
tem_df.groupby("match_id")["batsman_runs"].sum().sort_values(ascending =False).head(1).values[0]
```

```python
# 6: create a function that can return the highest score of any batsman.
def highest(batsman):
    tem_df = ipl[ipl["batsman"] == batsman]
    return tem_df.groupby("match_id")["batsman_runs"].sum().sort_values(ascending =False).head(1).values[0]

highest("CH Gayle")
```
