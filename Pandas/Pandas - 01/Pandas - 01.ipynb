## Missing Values

```python
# isnull ---> returns boolean series for checking if each item has value or it is a null value.
temp= pd.Series([1,2,np.nan,4,5,np.nan,np.nan,8,9,10])
temp
```

```python
temp.isnull() # gives boolean series
temp[temp.isnull()].size # 1st method to count the number of missing values.
```

```python
# 2nd method for missing values getting.
miss_values = temp.isnull().sum()
miss_values= int(miss_values)
miss_values
```

```python
# dropna # ---> drops all the missing values from the series and give a new series (dtype will be float64).
temp.dropna()
```

```python
# fillna # replace the missing value with the value provided in parentheisis and gives a new series.
temp.fillna(0)
```

```python
# you can also replace the missing value with any custom logic
# i.e. replace all the missing values with the mean of the data.
temp.fillna(np.round(temp.mean(),2))
```

```python
# q. find the matches where virat kohli outs at 49 0r 99
# 1st method.
vk[(vk == 49) | (vk == 99)] # using conditioning 
```

```python
# isin # --> return the boolean series for : "elements are in the series or not that we provided  ".
# 2nd method for question
vk[vk.isin([49,99])]
```

```python
vk[vk.isin([49,59,69,79,89,99])]
```
