## MultiIndex Series

```python
# can we have multiple index? let's try
index_val =[("cse",2019),("cse",2020),("cse",2021),("cse",2022),("ece",2019),("ece",2020),("ece",2021),("ece",2022)] # list of tuples of indicies
a = pd.Series([1,2,3,4,5,6,7,8],index=index_val)
# its working.
a[("cse",2021)]
a[("ece",2022)]
```

```python
# The problem:
a["cse"] 
# # i want to fetch all the rows in which cse data is stored but error because there is no index like "cse" it is combined ("cse",year) if we want to fetch data we have to provide all indicies in the tuple to fetch the result. this is the problem in this approach.

```

```python
# The solution ---> creating multiindex series -->uses multiindexing (also known as Hierarchical Indexing)
# multiple index levels within a single index
```

```python
#how to create multiindex object?
# 1. pd.MultiIndex.from_tuples()
index_val =[("cse",2019),("cse",2020),("cse",2021),("cse",2022),("ece",2019),("ece",2020),("ece",2021),("ece",2022)] # list of tuples of indicies
multiindex = pd.MultiIndex.from_tuples(index_val)
multiindex # multiindex object
```

```python
type(multiindex) # its type is Pandas.multiindex object
```

```python
# levels attribute
multiindex.levels # it decouples both the indicies given in a tuple and gives result in a frozen list
```

```python
multiindex.levels[0] # gives the 0th level of indexing --> branch
```

```python
multiindex.levels[1] # gives the 1st  level of indexing --> Year
```

```python
# 2: by using pd.MultiIndex.from_product()
pd.MultiIndex.from_product([["cse","ece"],[2019,2020,2021,2022]]) # passing a 2d list
# it will do Cartesian Product between Set A (list 1) and Set B (list 2)
```

```python
# creating a series with multiindex object
s = pd.Series([1,2,3,4,5,6,7,8],index=multiindex)
s
```

```python
# how to fetch items from such a series.
# fetching 4:
s[("cse",2022)]
```

```python
# fetch all the rows that have cse
s["cse"] # this we can access but in the above approach we cannot do this type of indexing
# we can access the branches of the tree.(Heirarchical Indexing)
s["ece"]
```

```python
# a logical question to ask?
# what is its dimension?

# this series is 2d because for fetching any value we have to provide two pieces of information --> branch ---> year
```

```python
# then why we not use dataframe?
# actually we convert this 2d series into a dataframe.
# unstack() Function
temp =s.unstack() # converts multiIndex series (2d) into dataframe
temp   
# it converts the level_01 index  into columns and level_0 index remains the index of rows of the dataframe.
# we can fetch same data by giving two pieces of information.
```

```python
# stack() function ---> converts dataframe into MultiIndex series (2d)

temp.stack() # --> # it converts the columns into level_01 index and rows indexes become level_0 indexes for the series.
```

```python
# Then what was the point of multiindex series?
# basically we represent high dimension data (2d) using a lower dimension object Series (1d)
# we can represent any higher dimension data like 3d,4d...nd using lower dimension object Dataframe (2d)
```

```python
temp_ser = temp_df.unstack() 
temp_ser# if we unstack again the level_0 index become level_02 column and no rows left only columsn so dataframe converted into series.
```

```python
temp_df2 = temp_df.stack() # level_01 colum becomes level_01 indexes
temp_df2
```

```python
temp_df2.stack() # level_0 column become level_02 index(row) and no cloumns left thats why dataframe becomes series.
```

## Wide Vs Long Data

**Wide Data Format**: Wide data format is where we have a single row for every data point with multiple columns to hold the values of various attributes.


**Long Data Format:** Long data format is where, for each data point we have as many rows as the number of attributes and each row contains the value of particular attribute for a given data point.

```python
# melt() convert wide data format into long data format
pd.DataFrame({
    "cse": [120]
}).melt()
```

```python
# var_name and value_name parameters
temp_df = pd.DataFrame({"cse":[120],"ese":[100],"mech":50})
temp_df # wide data
```

```python
temp_df.melt(var_name="branch",value_name="num_students") # long data
```

```python
# id_vars --> parameter ---> makes a specific column --> id (like index) and does not make it row (means skip that specific column)
pd.DataFrame({
    "branch":["cse","ece","mech"],
    "2020": [100,150,60],
    "2021": [120,130,80],
    "2022": [150,140,70]
}).melt(id_vars= ["branch"],var_name="year",value_name="num_students")
```
