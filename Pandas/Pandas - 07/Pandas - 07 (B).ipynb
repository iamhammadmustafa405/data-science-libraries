## **Vectorized String Operations**

```python
import numpy as np
import pandas as pd
```

```python
# what are vectorized operations?
a = np.array([1,2,3,4])
a*4 # this operations is applied to all the elements of the array these type of operations are called vectorized operations.
```

```python
# problem in vectorized  operation in vanilla python
s = ["cat","hat",4,"rat"]
[i.startswith("c") for i in s] # this code will crack because int object has no attribute startswith this is the problem in vectorized operation in vanilla python.
```

```python
# how pandas solve this issue?
s = pd.Series(["cat","hat",4,"rat"])
# .str ---> string accessor ---> when we have to use the string type functions we always have to use this accesor
s.str.startswith("c") # it igonores the 4 and treat as NaN value for this operation
# it is fast and optimized.
```

### **Common Functions**

```python
#lower
titanic["Name"].str.lower()
```

```python
# upper()
titanic["Name"].str.upper()
```

```python
# capitalize()
titanic["Name"].str.capitalize()
```

```python
# title()
titanic["Name"].str.title()
```

```python
# len() --. count the no.  of characters of the a string in series.
titanic["Name"].str.len()
# fetch the name of the passenger whose name is larges from all.
index_of_largest_name =titanic["Name"].str.len().sort_values(ascending=False).head(1).index[0]
titanic["Name"][index_of_largest_name]
```

```python
# another method
large_name = titanic["Name"].str.len().max()
titanic[titanic["Name"].str.len() == large_name]["Name"].values[0]
```

```python
# strip ---> remove all the leading and trailing spaces of the string.
"           Hammad         ".strip()
```

```python
# in Pandas
titanic["Name"].str.strip() # ---> remove all unwanted spaces from the series
```

```python
# split() -> split the string into parts on the basis of provided element.
titanic["Name"].str.split(" ") # split the Name column on the basis of "space".
```

```python
# get(index) --> get the nth index str element from the string after the split
titanic["Name"].str.split(" ").str.get(0)
```

```python
# question:
# Add 3 more columns to the dataframe Named: First Name --> Sr.Name/last Name -->Title
titanic["last_name"] = titanic["Name"].str.split(",").str.get(0) # fetched the Sr.name/last Name of all the passangers
titanic["last_name"]
```

```python
# n parameter ---> important # --> expand parameter make dataframe of columns in which splitting is performed
# again spliting
titanic[["title","first_name"]] = titanic["Name"].str.split(",").str.get(1).str.strip().str.split(" ",n=1,expand=True) # n=1 means only one split for the first space for the next spaces there is no split
titanic.head(3)
```

```python
# Questions 2: How many Mr. and Miss. are in titanic
titanic["title"].value_counts()
```

```python
# replace() ---> replace a string with another provided value.
# q: replace the Ms.  & Mlle. (Miss) value with Miss. because they are same.
titanic["title"] = titanic["title"].str.replace("Ms.","Miss.")
titanic["title"] = titanic["title"].str.replace("Mlle.","Miss.")
titanic["title"].value_counts().sort_values(ascending=False)
```

## **Filtering**

```python
#startswith/endswith
titanic[titanic["Name"].str.startswith("A")] # all the passangers whose name starts with "A"
```

```python
# endswith()
titanic[titanic["Name"].str.endswith("A")]
```

```python
# isdigit/isaplha()
titanic[titanic["first_name"].str.isdigit()]
```

### **Applying Regex (Regular Expressions)**

```python
# contains(case) --- > for searching specific part of the string 
#case parameter means case sensitivity True or False.
#  Q 1: search if the name of any passanger contains "john" or not.
titanic[titanic["first_name"].str.contains("john",case=False)]["Name"]
```

```python
# Q 2: find last_name with start and end char vowel.
# in this question we have to write regular expression. (Must learn the Regex First)
titanic[titanic["last_name"].str.contains("^[aeiouAEIOU].+[aeiouAEIOU]$")]
```

```python
# Q 3: find last_name with start and end char consonant.
titanic[titanic["last_name"].str.contains("^[^aeiouAEIOU].+[^aeiouAEIOU]$")]
```

```python
# slicing
titanic["Name"].str[:4] # fetch 4 characters of every string object in series.
titanic["Name"].str[::2] # fetch alternate characters of every string in series.
```

```python
# Q: Reverse the Name of every passangers in the dataset
titanic["Name"].str[::-1]
```
