```python
# Q 1: Find all the final winner
finals =ipl[ipl["MatchNumber"] == "Final"]
winners = finals[["Season","WinningTeam"]]
print(winners)
```

```python
# in one line
ipl[ipl["MatchNumber"] =="Final"][["Season","WinningTeam"]] # ---> this is professional way of writing logic.
```

```python
# Q 2: How many Super Over finishes have occured?
# Professional Way
result= int(ipl[ipl["SuperOver"] == "Y"]["SuperOver"].count())
print(f"{result} Super Over Occurs in Ipl History.")
```

```python
# 2nd method
ipl[ipl["SuperOver"] =="Y"].shape[0]
```

```python
# Q 3: How many matches has csk won in kolkata.
# My Logic
csk_won = ipl[ipl["WinningTeam"]=="Chennai Super Kings"]
result = csk_won[csk_won["City"] =="Kolkata"].shape[0]
print(f"CSK Won {result} Matches in Kolkata in Ipl History.")
```

```python
# Professional Method
ipl[(ipl["City"] =="Kolkata") & (ipl["WinningTeam"] == "Chennai Super Kings")].shape[0]
```

```python
#  Q 4: Toss winner is match winner in percentage
total= ipl.shape[0]
toss_winner =ipl[ipl["TossWinner"] == ipl["WinningTeam"]][["TossWinner","WinningTeam"]].shape[0]
percentage= (toss_winner/total) *100
percentage= np.round(percentage,2)
print(f"The Chances of Match Winning after Winning the toss are: {percentage}%")
```

```python
# Q 5: Movies with rating higher than 8 and vote>10000
movies[(movies["imdb_rating"] > 8)  & (movies["imdb_votes"] >10000)].shape[0]
```

```python
# Q 6: Action movies with rating higher than 7.5 
movies[(movies["genres"].str.split("|").apply(lambda x: "Action" in x)) & (movies["imdb_rating"] > 7.5)]
# This is the logic we have to build for making Machine Learning / Data Science Models.
```
