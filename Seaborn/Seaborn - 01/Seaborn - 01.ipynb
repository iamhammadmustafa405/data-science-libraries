```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
```

### Scatter Plot

```python
tips = pd.DataFrame({
    "total_bill" : [16.99,10.34,21.01,23.68,24.59,34.56,23.67,28.09,29.45,33.33,36.89],
    "tip": [1.01,1.66,3.50,3.31,3.61,3.71,3.89,4.00,4.02,4.06,4.23],
    "sex": ["Female","Male","Female","Male","Female","Male","Male","Female","Male","Female","Male"],
    "smoker": ["Yes","No","Yes","Yes","No","No","No","Yes","No","Yes","No"],
    "day": ["Sun","Sun","Sun","Sun","Sun","Mon","Mon","Tue","Tue","Fri","Sat"],
    "time": ["Dinner","Dinner","Dinner","Dinner","Dinner","Breakfast","Dinner","Lunch","Dinner","Breakfast","Lunch"],
    "size": [2,3,3,2,4,4,5,3,6,2,3]
})
tips["size"] = tips["size"] * 100
tips
```

```python
# scatterplot ---> sns.scatterplot ---> axes level function ---> Rectangular shaped (default)
# hue , style, size --> adding additional information in the plot
sns.scatterplot(data=tips,x="total_bill", y="tip",hue="sex",style="time",size="size")
# 5 column information is presented in plot
```

```python
# sns.relplot() -> general function to plot different plots by changing the kind 
# the axes level functions comes under the figure level functions.

# scatter plot using figure level function --> sns.relplot() -> Square shaped (default)
# data , x, y, kind ---> parameter

sns.relplot(data=tips,x="total_bill", y = "tip", kind="scatter",hue="sex",style="time",size="size")
```

### Line Plot

```python
gap =px.data.gapminder() # dataset in plotly
temp_df =gap[gap["country"] =="Pakistan"]
temp_df
```

```python
# line plot -> sns.lineplot --> axes level function
sns.lineplot(data=temp_df, x='year', y='lifeExp')
```

```python
# using sns.relplot() --> figure level function (suggested by seaborn library)
sns.relplot(data=temp_df,x="year", y="lifeExp",kind="line")
```

```python
# hue ,style, size --> paramters
temp_df = gap[gap["country"].isin(["Pakistan","Germany","Brazil"])]
temp_df
```

```python
# plotting graph of year vs life_expentency of different countries
sns.lineplot(data=temp_df,x="year",y="lifeExp",hue="country",style="continent",size="iso_num")
```

```python
# using relplot()
sns.relplot(kind="line",data=temp_df,x="year",y="lifeExp",hue="country",style="continent",size="iso_num")
```

```python
# facet plot --> for creating subplots, in a single figure, of different categories in the data using relplot()
# --> figure level function -> work only with sns.relplot()
# it will not work with scatterplot and lineplot.
sns.relplot(kind="scatter",data=tips,x="total_bill",y="tip",col="sex")
```

```python
sns.relplot(kind="scatter",data=tips,x="total_bill",y="tip",row ="sex")
```

```python
# make using rows and cols
sns.relplot(kind="scatter",data=tips,x="total_bill",y="tip",row="day",col="sex")
# row -> day -> 5 days in data , col -> sex -> 2 genders in data
# figure has row * col = 5*2 = 10 plots.
```

```python
# col_wrap -> parameter -> wrap graphs and shift other  to next row
sns.relplot(data=gap, x="lifeExp", y="gdpPercap",kind="scatter",col="country",col_wrap=3)
```

```python
# figure level -> displot()
# axes level -> histplot(),kdeplot(),rugplot()
```

```python
# univariate histogram
sns.histplot(data=tips,x="total_bill")
```

```python
# displot()
sns.displot(data=tips,x="total_bill",kind="hist")
```

```python
# bins parameter --> no. of bins
sns.displot(data=tips,x="total_bill",kind="hist",bins=10)
```

```python
# histogram on categorical column -> countplot
# it will assign discrete bins automatically
sns.displot(kind="hist",data=tips,x="day")
```

```python
# hue parameter
sns.displot(kind="hist",data = tips,x="tip",hue="sex")
```

```python
# for only outline of bar -> element ="step" parameter
sns.displot(kind="hist",data=tips,x="tip",hue="sex",element="step")
```

## **KDE (Kernal Density Estimation) Plot**

```python
#kde plot --> rather than discrete , a KDE smooths the observations witha a function called Gussian Kernal, producing a continuous density estimate
# sns.kdeplot() --> axes level
sns.kdeplot(data=tips,x="total_bill") # y axis --> density of probability of the respective x axis value. 
```

```python
# figure level -> displot()
sns.displot(kind="kde",data=tips,x="total_bill")
```

```python
# hue -> fill -> parameter
sns.displot(kind="kde",data=tips,x="total_bill",hue="sex")
```

```python
sns.displot(kind="kde",data=titanic,x="Fare", hue="Sex",fill=True) # fill -> fill the area under curve
```

## **Rug Plot**

```python
# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an obtrusive way.

# sns.rugplot()

sns.kdeplot(data=titanic,x="Fare")
sns.rugplot(data=titanic,x="Fare")
```

```python

# Bivariate Histogram
# A Bivariate Histogram bins the data within rectangles that tile the plot.
# and then shows the count of observations within each rectangle with the fill color. 
# Interpretation : ---> darker the color of the tile -> higesht frequency count of that particular combination
sns.histplot(data=tips,x="total_bill",y="tip")
```

```python
sns.displot(kind="hist",data=tips,x="total_bill",y="tip")
```

```python
# Bivariate KDE Plot
# a bivariate KDE plot smoothes the (x,y)  observations with a 2D Guassian Kernal Function
# Interpretation: --> the more closer the line --> the higher the frequency of that particular combination of (x,y).
sns.kdeplot(data=tips,x="total_bill",y="tip")
```

# **3. Matrix Plot**

Heatmap

ClusterMap

```python
# heatmap() -> axes level function
# there is no corresponding figure level function to heatmap
temp_df = gap.pivot(index="country",columns="year",values="lifeExp") #wide format data
temp_df
```

```python
# axes level function can be customized like in matplotlib
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
plt.title("Heatmap of Life Expectancy of Worldwide Countries")
```

```python
temp_df = gap[gap["continent"] =="Europe"].pivot(index="country",columns="year",values="lifeExp")
temp_df
```

```python
# annot -> give a number to each tile
# linewidths-> seperation btw the tiles
plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidths=0.5,cmap="viridis")
plt.title("Heatmap of Life Expectancy of European Countries")
```
