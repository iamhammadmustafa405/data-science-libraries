# **Seaborn - 02**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
iris = px.data.iris()
iris
```

```python
tips = px.data.tips()
tips
```

# **4.Categorical Plots**

### **Categorical Scatter Plots**

##### **1). Strip Plot**

```python
# strip plot -> categorical vs numerical
# axes level function
# jitter = False parameter -> real graph (by default seaborn add noise to seperate the points for reaability that not the real points in real place)
sns.stripplot(data=tips,x="day",y="total_bill",jitter=False)
```

```python
# figure level 
# catplot
sns.catplot(data=tips,x="day",y="total_bill",kind="strip")
```

```python
# jitter and hue
sns.catplot(data=tips,x="day",y="total_bill",kind="strip",hue="sex",jitter=0.2)
```

#### **Swarm Plot**

```python
# swarm plot -> axes level
# same like strip but uses internal algorithm to adjust the data in a way in which the idea for the distribution can be easily understood.
# for small dataset
sns.swarmplot(data=tips,x="day",y="total_bill",hue="sex")
```

```python
# using catplot -> figure level
sns.catplot(data=tips,x="day",y="total_bill",kind="swarm",hue="sex")
```

## **Categorical Distribution Plots**

### **Box Plot**

```python
#box plot -> numerical col -> 5 number summary of data
sns.boxplot(data=tips,y="total_bill")
```

```python
# box plot -> categorical vs numerical (seaborn feature)
sns.boxplot(data=tips,x="sex",y="total_bill")
```

```python
sns.boxplot(data=tips,x="day",y="total_bill",hue="sex")
```

```python
# using catplot
sns.catplot(data=tips,x="day",y="total_bill",kind="box",hue="sex")
```

#### **Violin Plot**

```python
# violin plot = (box plot + Kde Plot) -> symmetrical
sns.violinplot(data=tips,x="day",y="total_bill")
```

```python
# catplot
sns.catplot(data=tips,x="day",y="total_bill",kind="violin",hue="sex")
```

```python
# Split -> remove the symmetricity of the garph
sns.catplot(data=tips,x="day",y="total_bill",kind="violin",hue="sex",split=True)
```

## **Categorical Estimate Plot (Measuring Central Tendency/Aggregate Value)**

#### **Bar plot**

```python
# bar plot -> calculate estimate of some categorical column with numerical column (default is mean)

sns.barplot(data=tips,x="sex",y="total_bill")
```

```python
# catplot
# errobar = none
# When there are multiple observations in each category. It also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars.
sns.catplot(data=tips,x="sex",y="total_bill",kind="bar",errorbar=None)
```

```python
# hue , estimator -> estimation on numerical column
sns.catplot(kind="bar",data=tips,x="sex",y="total_bill",hue="smoker",estimator=np.std)
```

```python
sns.catplot(kind="bar",data=tips,x="sex",y="total_bill",hue="smoker",estimator=np.min)
```

## **Point Plot**

```python
sns.barplot(data=tips,x="sex",y="total_bill",errorbar=None,hue="smoker") # gives bar to show estimate
```

```python
# point plot -> it same like bar but insted of bars it  gives the difference of categories by connecting points to make line
sns.pointplot(data=tips,x="sex",y="total_bill",hue="smoker",errorbar=None) # gives difference between the estimate of the bars
```

### **Count Plot**

```python
#count plot -> counts the given categorical column
# A special case of bar plot in which when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical.rather than quantitative variable

sns.countplot(data=tips,x="sex")
```

```python
# faceting with catplot
# row, col, column_wrap etc.
sns.catplot(kind="box",data=tips,x="sex",y="total_bill",col="smoker",row="day")
```

## **Regression Plots**

### **Reg Plot**

In the simpleset invocation,both (reg + lm plot) draw a scatterplot of two variables x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression
shadow is confidence interval.

```python
# regplot --> scatterplot + linear regression line
# hue parameter not available
# axes level function
sns.regplot(data=tips,x="total_bill",y="tip")
```

```python
# figure level function -> lmplot()
# do not need to provide kind parameter
# hue parameter available
sns.lmplot(data=tips,x="total_bill",y="tip",hue="sex")
```

## **Residual Plot**

```python
# residplot -> error of regression model in plot
sns.residplot(data=tips,x="total_bill",y="tip")
```

## **MultiGrid Plots**

## **FacetGrids --> Another way to Use Facet**

```python
# 1st. using figure level function -> relplot displot catplot lmplot
# Generally 99% this method is used for facets
sns.catplot(kind="violin",data=tips,x="sex",y="total_bill",row="time",col="day")
```

```python
# 2nd. using FacetGrid()
g = sns.FacetGrid(data=tips,row="time",col="day") # facetgrid object
g.map(sns.boxplot,"sex","total_bill")  # g.map(axes function , x, y) 
# Generally it not used so much
```

## **Pair Plot -> Useful**

```python
# pairplot -> plot scatter plot btw all combination of the numerical column
sns.pairplot(iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width','species']])
```

```python
sns.pairplot(iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width','species']],hue="species")
```

## **PairGrid**

```python
# same pairplot using pairgrid
g = sns.PairGrid(data=iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width','species']])
g.map(sns.scatterplot)
```

```python
# map_diag & map_offdiag
g = sns.PairGrid(data=iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width','species']],hue="species")
g.map_diag(sns.histplot) # -> plots in diagonal ( same column plots)
g.map_offdiag(sns.scatterplot) # -> plots in off diagonal (different column plots)
```

```python
g = sns.PairGrid(data=iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width','species']],hue="species")
g.map_diag(sns.violinplot)
g.map_offdiag(sns.kdeplot) 
```

```python
# map_diag -> map_lower -> map_upper, vars
g = sns.PairGrid(data=iris,vars=['sepal_length', 'sepal_width'])
g.map_diag(sns.histplot)
g.map_lower(sns.histplot)
g.map_upper(sns.scatterplot)
```

## **Joint Grid**

```python
# joint Grid
g = sns.JointGrid(data=tips,x="total_bill",y="tip")
g.plot(sns.scatterplot,sns.histplot)
```

```python
g = sns.JointGrid(data=tips,x="total_bill",y="tip")
g.plot(sns.kdeplot,sns.boxplot)
```
